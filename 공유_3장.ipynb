```python
# -*- coding: utf-8 -*-
import json
import pandas as pd
import matplotlib
#matplotlib.use('Agg')  # GUI 없이 실행
import matplotlib.pyplot as plt
from matplotlib import rc
import os

# 한글 폰트 설정
#rc('font', family='Malgun Gothic')
#plt.rcParams['axes.unicode_minus'] = False

# 현재 스크립트 디렉토리
script_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in globals() else os.getcwd()
json_file = os.path.join(script_dir, '공유_year_word_frequency_papers_only.json')

print(f"JSON 파일 경로: {json_file}")

# JSON 불러오기
with open(json_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

print("데이터 로드 완료")

# 데이터프레임 변환
rows = []
for decade, words in data.items():
    for word, count in words.items():
        rows.append({'decade': int(decade), 'word': word, 'count': count})

df = pd.DataFrame(rows)

print(f"데이터프레임 생성 완료: {len(df)}개 행")

# 분석 대상 키워드
targets = ["일제", "일제강점기", "일제하", "근대", "식민지", "동아시아"]

print(f"\n분석 대상 키워드: {targets}")

# 키워드별 총 빈도 확인
print("\n키워드별 총 빈도:")
for keyword in targets:
    keyword_df = df[df["word"] == keyword]
    if len(keyword_df) > 0:
        total = keyword_df["count"].sum()
        print(f"  {keyword}: {total}회")
    else:
        print(f"  {keyword}: 데이터 없음")

df_filtered = df[df["word"].isin(targets)]

if len(df_filtered) == 0:
    print("\n오류: 필터링된 데이터가 없습니다.")
    exit(1)

pivot = df_filtered.pivot_table(index="decade", columns="word",
                                values="count", fill_value=0)

print(f"\n피벗 테이블 생성 완료")
print(f"연대 범위: {pivot.index.min()}년대 ~ {pivot.index.max()}년대")
print(f"\n피벗 테이블 미리보기:")
print(pivot)

# 그래프
plt.figure(figsize=(14, 8))

# 색상 팔레트 설정 (6개 키워드용)
colors = ['#e74c3c', '#3498db', '#9b59b6', '#2ecc71', '#f39c12', '#1abc9c']
markers = ['o', 's', '^', 'd', 'v', 'p']

for i, col in enumerate(pivot.columns):
    if col in targets:  # 존재하는 키워드만 그리기
        plt.plot(pivot.index, pivot[col], marker=markers[i % len(markers)], 
                 label=col, linewidth=2.5, markersize=7, 
                 color=colors[i % len(colors)])

plt.title("일제·근대·식민지 관련 키워드의 시대별 변화 추이", 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel("연대", fontsize=14, fontweight='bold')
plt.ylabel("등장 빈도", fontsize=14, fontweight='bold')
plt.legend(loc='best', fontsize=11, framealpha=0.9)
plt.grid(True, alpha=0.3, linestyle='--')

output_path = os.path.join(script_dir, '공유_keywords_modern_graph.png')
plt.tight_layout()
plt.savefig(output_path, dpi=300, bbox_inches='tight')
plt.show()
print(f"\n그래프 저장 완료: {output_path}")
plt.close()

print("완료!")

```

```python
# -*- coding: utf-8 -*-
import json
import pandas as pd
import matplotlib
#matplotlib.use('Agg')  # GUI 없이 실행
import matplotlib.pyplot as plt
from matplotlib import rc
import os

# 한글 폰트 설정
#rc('font', family='Malgun Gothic')
#plt.rcParams['axes.unicode_minus'] = False

# 현재 스크립트 디렉토리
script_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in globals() else os.getcwd()
json_file = os.path.join(script_dir, '공유_year_word_frequency_papers_only.json')

print(f"JSON 파일 경로: {json_file}")

# JSON 불러오기
with open(json_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

print("데이터 로드 완료")

# 데이터프레임 변환
rows = []
for decade, words in data.items():
    for word, count in words.items():
        rows.append({'decade': int(decade), 'word': word, 'count': count})

df = pd.DataFrame(rows)

print(f"데이터프레임 생성 완료: {len(df)}개 행")

# 전체 상위 10개 키워드
top10 = df.groupby("word")["count"].sum().sort_values(ascending=False).head(10).index.tolist()

print(f"\n상위 10개 키워드:")
for i, word in enumerate(top10, 1):
    total = df[df["word"] == word]["count"].sum()
    print(f"{i}. {word}: {total}회")

df_top = df[df["word"].isin(top10)]

pivot = df_top.pivot_table(index="decade", columns="word",
                           values="count", fill_value=0)

print(f"\n피벗 테이블 생성 완료")
print(f"연대 범위: {pivot.index.min()}년대 ~ {pivot.index.max()}년대")

# 그래프
plt.figure(figsize=(14, 8))

# 색상 팔레트 설정 (10개 키워드용)
colors = ['#e74c3c', '#3498db', '#9b59b6', '#2ecc71', '#f39c12', 
          '#1abc9c', '#e67e22', '#34495e', '#16a085', '#c0392b']

for i, col in enumerate(pivot.columns):
    plt.plot(pivot.index, pivot[col], marker='o', label=col, 
             linewidth=2.5, markersize=6, color=colors[i % len(colors)])

plt.title("상위 10개 키워드의 시대별 변화 추이", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("연대", fontsize=14, fontweight='bold')
plt.ylabel("등장 빈도", fontsize=14, fontweight='bold')
plt.legend(loc='best', fontsize=10, framealpha=0.9, ncol=2)
plt.grid(True, alpha=0.3, linestyle='--')

output_path = os.path.join(script_dir, '공유_top10_keywords_graph.png')
plt.tight_layout()
plt.savefig(output_path, dpi=300, bbox_inches='tight')
plt.show()
print(f"\n그래프 저장 완료: {output_path}")
plt.close()

print("완료!")

```
